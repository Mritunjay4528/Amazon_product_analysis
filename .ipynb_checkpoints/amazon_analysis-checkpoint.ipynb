```python
import pandas as pd
```

```python
df = pd.read_csv("amazon_products.csv")

df.head()
```

# rank top 100 products by volume (quantity)

```python
top_volume = df.sort_values(
    by="Reviews",
    ascending=False
)

top_volume.head(100)
```

```python
top_volume.head(100).to_csv(
    "top_100_by_volume.csv",
    index=False
)
```

# rank top 100 products by value

```python
df["Estimated_Value"] = df["Reviews"] * df["Price"]
```

```python
top_value = df.sort_values(
    by="Estimated_Value",
    ascending=False
)

top_value.head(100)
```

```python
top_value.head(100).to_csv(
    "top_100_by_value.csv",
    index=False
)
```
