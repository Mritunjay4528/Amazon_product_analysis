```python
all_products = []

for file_name in html_files:

    # Skip non-html files
    if not file_name.endswith(".html"):
        continue

    file_path = os.path.join("raw_html", file_name)

    # Open HTML file
    with open(file_path, "r", encoding="utf-8") as file:

        html_content = file.read()

    # Create soup
    soup = BeautifulSoup(html_content, "lxml")

    # Find all products
    products = soup.find_all(
        "div",
        attrs={"data-asin": True}
    )

    print(f"\nProcessing: {file_name}")
    print(f"Products Found: {len(products)}")

    # Loop through products
    for product in products:

        try:

            # ASIN
            asin = product.get("data-asin")

            # Product Name
            name_tag = product.find(
                "div",
                class_="_cDEzb_p13n-sc-css-line-clamp-3_g3dy1"
            )

            # Backup selector
            if not name_tag:

                name_tag = product.find(
                    "div",
                    class_="_cDEzb_p13n-sc-css-line-clamp-4_2q2cc"
                )

            # Skip if no name
            if not name_tag:
                continue

            name = name_tag.text.strip()

            # Reviews
            reviews_tag = product.find(
                "span",
                class_="a-size-small"
            )

            if reviews_tag:

                reviews = reviews_tag.text.strip()

            else:

                reviews = "0"

            # Price
            price = None
            
            all_spans = product.find_all("span")
            
            for span in all_spans:
            
                text = span.text.strip()
            
                if "₹" in text:
            
                    price = text
                    break
            
            # Skip if no price
            if not price:
                continue

            # Print extracted data
            print(asin)
            print(name)
            print(reviews)
            print(price)
            print("-" * 50)

            # Store product
            product_data = {
                "ASIN": asin,
                "Product_Name": name,
                "Reviews": reviews,
                "Price": price
            }

            all_products.append(product_data)

        except Exception as e:

            print("Error:", e)
```

# final working code

```python
all_products = []

for file_name in html_files:

    # Skip non-html files
    if not file_name.endswith(".html"):
        continue

    file_path = os.path.join("raw_html", file_name)

    # Open HTML file
    with open(file_path, "r", encoding="utf-8") as file:

        html_content = file.read()

    # Create soup
    soup = BeautifulSoup(html_content, "lxml")

    # Find all products
    products = soup.find_all(
        "div",
        attrs={"data-asin": True}
    )

    print(f"\nProcessing: {file_name}")
    print(f"Products Found: {len(products)}")

    # Loop through products
    for product in products:

        try:

            # ASIN
            asin = product.get("data-asin")

            # Product Name
            name_tag = product.find(
                "div",
                class_="_cDEzb_p13n-sc-css-line-clamp-3_g3dy1"
            )

            # Backup selector
            if not name_tag:

                name_tag = product.find(
                    "div",
                    class_="_cDEzb_p13n-sc-css-line-clamp-4_2q2cc"
                )

            # Skip if no name
            if not name_tag:
                continue

            full_name = name_tag.text.strip()

            # Clean product name
            name = full_name.split(",")[0]

            name = name.split("|")[0]

            name = " ".join(name.split()[:5])

            # Reviews
            reviews_tag = product.find(
                "span",
                class_="a-size-small"
            )

            if reviews_tag:

                reviews = reviews_tag.text.strip()

                reviews = reviews.replace(",", "")

            else:

                reviews = "0"

            # Price
            price = None

            all_spans = product.find_all("span")

            for span in all_spans:

                text = span.text.strip()

                if text.startswith("₹"):

                    price = text
                    break

            # Skip if no price
            if not price:
                continue

            # Clean price
            price = (
                price.replace("₹", "")
                     .replace(",", "")
            )

            # Print extracted data
            print(asin)
            print(name)
            print(reviews)
            print(price)
            print("-" * 50)

            # Store product
            product_data = {
                "ASIN": asin,
                "Product_Name": name,
                "Reviews": int(reviews),
                "Price": float(price)
            }

            all_products.append(product_data)

        except Exception as e:

            print("Error:", e)

print("\nTotal Products Extracted:", len(all_products))
```
